## Mission 9：加入缺失值與異常值
#### Knowledge Point

###### 使用 .loc 按列與欄位索引賦值
###### np.nan 代表缺失值（Missing）
###### 索引從 0 開始，第 5 列對應 student_id 6，第 17 列對應 student_id 18
###### 故意設置極端值（15 小時螢幕使用時間）作為異常值示範
#### Markdown Cell

###### 在真實資料中，常會遇到缺失或極端的數據。此步驟我們故意引入兩筆 sleep_hours 的缺失值與一筆異常的 screen_time_hours，以便後續練習資料品質檢查。

###### Code Cell

```python

# 故意將第 5 列與第 17 列的 sleep_hours 設為缺失值
df.loc[[5, 17], "sleep_hours"] = np.nan

# 故意將第 8 列的 screen_time_hours 設為極端值 15.0（超出合理範圍）
df.loc[8, "screen_time_hours"] = 15.0

# 檢查這三筆資料是否真的被改變
print("兩筆睡眠缺失:")
display(df.loc[[5, 17], ["student_id", "sleep_hours"]])
print("一筆螢幕時間異常")
display(df.loc[[8], ["student_id", "screen_time_hours"]])
```

## Mission 11：篩選與比較資料
#### Knowledge Point

###### 建立布林條件（True/False）
###### 把條件儲存到變數中，再用來篩選 DataFrame
###### 單欄位、多欄位選取
###### 篩選缺失值、篩選極端值、篩選高分數
#### Markdown Cell

###### 我們將透過布林條件來挑出符合特定標準的資料列，例如：成績≥80、螢幕使用時間>10、有缺失值的睡眠時數等。先把條件放在變數中會讓程式更易閱讀與除錯。

#### Code Cell

```python
# 1️⃣ 單一欄位選取：取得 quiz_score 欄位（回傳 Series）
quiz_scores = df["quiz_score"]
print("quiz_score 前五筆：")
print(quiz_scores.head())

# 2️⃣ 多欄位選取：取得我們關心的六個欄位
selected_cols = df[
    ["student_id", "study_hours", "sleep_hours",
     "screen_time_hours", "attendance_rate", "quiz_score"]
]
print("\n選取六個欄位的前五筆：")
print(selected_cols.head())

# 3️⃣ 找出有缺失值的資料（以 sleep_hours 為例）
missing_sleep = df["sleep_hours"].isna()
missing_sleep_students = df[missing_sleep]
print("\n有缺失 sleep_hours 的學生（student_id）：")
print(missing_sleep_students[["student_id", "sleep_hours"]])

# 4️⃣ 找出螢幕使用時間超過 10 小時的資料
high_screen = df["screen_time_hours"] > 10
high_screen_students = df[high_screen]
print("\n螢幕使用時間 > 10 小時的學生：")
print(high_screen_students[["student_id", "screen_time_hours"]])

# 5️⃣ 找出小考成績 80 分以上的學生
high_score = df["quiz_score"] >= 80
high_score_students = df[high_score]
print("\nquiz_score >= 80 的學生人數：", high_score_students.shape[0])
print("前五筆高分學生：")
print(high_score_students[
    ["student_id", "quiz_score", "study_hours", "sleep_hours"]
].head())
```
